```python
# This scoring function provides an ESTIMATE of your course grade.
# Final calculations will be done by staff, using the posted algorithm.
#    ECE 3, UCSB
#    Katie Byl, Oct. 28, 2026


def course_grade(hw, lab, lecture, q1, q2, final_12, final_34, ec=0):
    """Estimate your course grade (all scores are percentages, 0-100).

    final_12 / final_34: your scores on the Units 1-2 and Units 3-4
    halves of the final exam. ec: extra-credit course points earned (0-15).

    Returns (grade, max_useful_ec, unused_ec):
      grade          -- estimated course percentage, including usable EC
      max_useful_ec  -- the most EC that could still raise your grade
      unused_ec      -- EC you earned that had nowhere to go
    """
    # 1. Blend: a quiz counts as the better of (original) or
    #    (average of the quiz and the matching half of the final).
    q1_eff = max(q1, (q1 + final_12) / 2)
    q2_eff = max(q2, (q2 + final_34) / 2)
    final = (final_12 + final_34) / 2

    # 2. Weighted scores: (weight in % of course, your score)
    parts = [(12, lecture), (12, lab), (12, hw),
             (19, q1_eff), (19, q2_eff), (26, final)]
    base = sum(w * s / 100 for w, s in parts)

    # 3. EC can raise any part by at most 15 points (max 100%),
    #    so it only fills in where you've lost points.
    headroom = sum(w * min(15, 100 - s) / 100 for w, s in parts)
    max_useful_ec = min(15, headroom)
    bonus = min(ec, max_useful_ec)
    unused_ec = ec - bonus

    return base + bonus, max_useful_ec, unused_ec


# Below, you can play with individual values
hw = 100
lab = 100
lecture = 50
q1 = 50
q2 = 80
final_12 = 80
final_34 = 50
ec = 15

grade, max_useful_ec, unused_ec = course_grade(
    hw, lab, lecture, q1, q2, final_12, final_34, ec=ec)
print('Quick Grade Estimator:')
print('- My grade would be:', round(grade, 2))
print('- The most EC that could help my grade is:', round(max_useful_ec, 2))
print('- The EC I earned that would be unused:', round(unused_ec, 2))

```
