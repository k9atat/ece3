```python
# (on the slides, this code is hidden: only its output shows)
# Shared look for the plots in this deck (hidden on slides)
import numpy as np, matplotlib.pyplot as plt
plt.rcParams.update({"font.size": 15, "axes.titlesize": 17, "axes.spines.top": False,
                     "axes.spines.right": False, "axes.edgecolor": "#888", "axes.labelcolor": "#333",
                     "xtick.color": "#444", "ytick.color": "#444", "figure.facecolor": "white",
                     "axes.facecolor": "white", "savefig.facecolor": "white"})
BLUE, ORANGE, GRAY, GREEN, RED, PURPLE = "#2a78d6", "#eb6834", "#b9b8b2", "#2da44e", "#d1242f", "#8250df"
```

```python
# (on the slides, this code is hidden: only its output shows)
# k-means from scratch, used by the plots below (hidden)
import pandas as pd
def assign(X, C):
    labels = np.zeros(len(X), dtype=int)
    for i in range(len(X)):
        labels[i] = np.argmin(np.linalg.norm(C - X[i], axis=1))
    return labels

def update(X, labels, k):
    return np.array([X[labels == j].mean(axis=0) for j in range(k)])

def cost(X, C, labels):
    return np.sum((X - C[labels]) ** 2)

def kmeans_mine(X, k, seed=0):
    rng = np.random.default_rng(seed)
    C = X[rng.choice(len(X), k, replace=False)].astype(float)
    for _ in range(100):
        lab = assign(X, C); Cn = update(X, lab, k)
        if np.allclose(Cn, C):
            break
        C = Cn
    return C, lab

COLS = np.array([BLUE, ORANGE, GREEN, PURPLE, RED])
def panel(ax, X, C, labels=None, title=""):
    if labels is None:
        ax.plot(X[:, 0], X[:, 1], "o", ms=6, color=GRAY, mec="#555")
    else:
        ax.scatter(X[:, 0], X[:, 1], s=36, c=COLS[labels], edgecolors="#333", linewidths=0.4)
    ax.scatter(C[:, 0], C[:, 1], s=420, marker="X", c=COLS[:len(C)], edgecolors="black", linewidths=2, zorder=5)
    ax.set_title(title, loc="left", fontsize=15); ax.set_xticks([]); ax.set_yticks([]); ax.set_aspect("equal")

pilots = pd.read_csv("data/pilot_lines.csv")
```

## Step 1 in code: assign _Not on Quiz 2..._

```python
def assign(X, C):
    labels = np.zeros(len(X), dtype=int)
    for i in range(len(X)):                      # every point
        dist = np.linalg.norm(C - X[i], axis=1)  # distance to every centroid
        labels[i] = np.argmin(dist)              # index of the nearest
    return labels
```

## Step 2 in code, and the whole loop _Not on Quiz 2..._

```python
def update(X, labels, k):
    return np.array([X[labels == j].mean(axis=0)
                     for j in range(k)])

def kmeans(X, k, seed=0):
    rng = np.random.default_rng(seed)
    C = X[rng.choice(len(X), k, replace=False)]
    while True:
        labels = assign(X, C)
        C_new = update(X, labels, k)
        if np.allclose(C_new, C):  # nothing moved
            return C, labels
        C = C_new

C, labels = kmeans(X, 3)
print(np.round(C, 2))
```

## What is k-means trying to do?

```python
# (on the slides, this code is hidden: only its output shows)
C = np.array([[0.0, 6.5], [1.0, 6.5], [6.5, 6.5]]); Js = []
for _ in range(8):
    lab = assign(X, C); Js.append(cost(X, C, lab)); C = update(X, lab, 3)
fig, ax = plt.subplots(figsize=(15, 3.2))
ax.plot(range(len(Js)), Js, "o-", ms=10, lw=3, color=BLUE)
ax.set_xlabel("round"); ax.set_ylabel("J")
ax.set_title("Assign and update can only lower J (or keep it), so the loop always stops", loc="left")
plt.tight_layout(); plt.show()
```

## When the hunch is wrong: k-means still answers

```python
# (on the slides, this code is hidden: only its output shows)
def best_of(X, k, tries=10):                  # best J over 10 seeds: these are not bad starts
    return min((kmeans_mine(X, k, seed=s) for s in range(tries)), key=lambda r: cost(X, *r))
rng = np.random.default_rng(7)
t1, t2, t3 = rng.uniform(0, np.pi, 100), rng.uniform(0, np.pi, 100), rng.uniform(0, 2 * np.pi, 150)
far = np.vstack([rng.normal([0, 0], 0.5, (12, 2)), rng.normal([3, 0], 0.5, (12, 2)), [[12, 0.0]]])
cases = [
    ("long, thin groups: cut across", np.vstack([np.c_[rng.uniform(0, 10, 80), rng.normal(0, 0.3, 80)],
                                                 np.c_[rng.uniform(0, 10, 80), rng.normal(2.2, 0.3, 80)]]), 2),
    ("a big group and a small one", np.vstack([rng.normal([0, 0], 1.6, (150, 2)), rng.normal([4.2, 0], 0.35, (20, 2))]), 2),
    ("two moons: not round", np.vstack([np.c_[np.cos(t1), np.sin(t1)], np.c_[1 - np.cos(t2), 0.5 - np.sin(t2)]])
                             + rng.normal(0, 0.07, (200, 2)), 2),
    ("a ring around a blob", np.vstack([np.c_[3 * np.cos(t3), 3 * np.sin(t3)] + rng.normal(0, 0.15, (150, 2)),
                                        rng.normal(0, 0.4, (50, 2))]), 2),
    ("one far point drags a centroid (○ = without it)", far, 2),
    ("pure noise: 4 'groups' anyway", rng.uniform(0, 6, (150, 2)), 4),
]
fig, axes = plt.subplots(2, 3, figsize=(15, 6.4))
for ax, (title, Xc, k) in zip(axes.flat, cases):
    C, lab = best_of(Xc, k)
    ax.scatter(Xc[:, 0], Xc[:, 1], s=16, c=COLS[lab], edgecolors="none")
    ax.scatter(C[:, 0], C[:, 1], s=260, marker="X", c=COLS[:k], edgecolors="black", linewidths=1.8, zorder=5)
    ax.set_title(title, loc="left", fontsize=13); ax.set_xticks([]); ax.set_yticks([])
    ax.set_aspect("equal", adjustable="datalim")
ax = axes.flat[4]; lab_far = best_of(far, 2)[1]
j = lab_far[-1]                               # the cluster that took the far point
ax.scatter(*far[:-1][lab_far[:-1] == j].mean(axis=0), s=260, marker="o", facecolors="none", edgecolors="black", linewidths=2, zorder=6)
plt.tight_layout(); plt.show()
```

# k-nearest neighbors: use the labels

```python
# (on the slides, this code is hidden: only its output shows)
# two labeled classes of clean, made-up data, and a simple k-nearest-neighbors (hidden)
rng = np.random.default_rng(11)
A = rng.normal([3, 3], 1.0, (25, 2)); B = rng.normal([6, 5.5], 1.0, (25, 2))
XL = np.vstack([A, B]); yL = np.array([0] * 25 + [1] * 25)
CL = np.array([BLUE, ORANGE]); NAMES = ["class A", "class B"]
def knn(Xtr, ytr, P, k):
    d = np.linalg.norm(Xtr - P, axis=1); near = np.argsort(d)[:k]
    return np.bincount(ytr[near], minlength=2).argmax(), near
```

## (If we have time...) Mission control: which pilot, or abort?

```python
# (on the slides, this code is hidden: only its output shows)
F = land[["v0", "fuel"]].to_numpy(float); sdF = F.std(axis=0)
def predict(pilot, pts, k=7):
    sub = land[land["pilot"] == pilot]; Xp = sub[["v0", "fuel"]].to_numpy(float) / sdF; yp = sub["outcome"].to_numpy()
    out = []
    for p in pts / sdF:
        near = np.argsort(np.linalg.norm(Xp - p, axis=1))[:k]
        vals, cnt = np.unique(yp[near], return_counts=True); out.append(vals[cnt.argmax()])
    return np.array(out)
gv, gf = np.meshgrid(np.linspace(-60, -5, 70), np.linspace(1000, 5000, 60))
G = np.column_stack([gv.ravel(), gf.ravel()])
gentle, saver = predict("gentle", G), predict("fuel saver", G)
plan = np.where(gentle == "gentle", 0, np.where(np.isin(saver, ["gentle", "hard"]), 1, 2)).reshape(gv.shape)
fig, axes = plt.subplots(1, 3, figsize=(16, 5.0), sharey=True)
for ax, pilot in zip(axes[:2], ["gentle", "fuel saver"]):          # the labeled data k-nearest neighbors looks up
    sub = land[land["pilot"] == pilot]
    for o, col in OCOL.items():
        m = sub["outcome"] == o
        if m.any():
            ax.scatter(sub["v0"][m], sub["fuel"][m], s=14, color=col, label=o)
    ax.set_title(f"{len(sub)} landings by the {pilot.upper()} pilot", loc="left", fontsize=14)
ax = axes[2]
ax.contourf(gv, gf, plan, levels=[-0.5, 0.5, 1.5, 2.5], colors=["#c9ecd1", "#fde3c8", "#f6c9cc"])
for lab, (x, y) in {"GENTLE pilot": (-22, 4300), "FUEL SAVER\n(hard landing)": (-22, 2600), "ABORT": (-50, 1800)}.items():
    ax.text(x, y, lab, ha="center", va="center", fontsize=14, weight="bold")
ax.set_title("k = 7 votes, for each pilot: the plan", loc="left", fontsize=14)
for a in axes: a.set_xlabel("velocity at 120 m (m/s)")
axes[0].set_ylabel("fuel at 120 m (kN·s)")
h = [plt.Line2D([], [], ls="", marker="o", ms=8, color=c) for c in OCOL.values()]
fig.legend(h, list(OCOL), loc="lower center", ncol=4, fontsize=13, frameon=False, bbox_to_anchor=(0.36, 0.0))
plt.tight_layout(rect=[0, 0.08, 1, 1]); plt.show()
```
