## Raw data: 50 bike-share ride times (made-up data)

```python
import numpy as np

# Imagine these are individual ride times (in minutes) from a bike-share app
np.random.seed(0)
raw_durations = np.random.randint(5, 20, size=50)    # every ride takes at least 5 minutes
raw_durations = raw_durations + [int(x) for x in (25*np.random.exponential(1, size=50))]   # plus a long tail
print(raw_durations)
```

## The same data, as a DataFrame

```python
import pandas as pd
df = pd.DataFrame({"ride_duration_min": raw_durations})
df.head(5)
```

## Summarize it: **Numerically** and **Visually**

```python
import matplotlib.pyplot as plt

print("\nSummary statistics:")
print(df["ride_duration_min"].describe())

# Compute specific measures
mean_val = df["ride_duration_min"].mean()
median_val = df["ride_duration_min"].median()
print(f"\nMean = {mean_val:.1f} min, Median = {median_val:.1f} min")

# Visualization
plt.figure(figsize=(7, 2.6))
plt.hist(df["ride_duration_min"], bins=15, edgecolor="black")
plt.xlabel("Ride Duration (minutes)")
plt.ylabel("Number of rides")
plt.yticks(range(0, 16, 2))
plt.title("Distribution of Ride Durations")
plt.show()
```

## Mean in Python

## Many means at once: the `axis` argument

```python
x1 = [98, 99, 100, 101, 102]
x2 = [80, 90, 100, 110, 120]
x3 = [60, 70, 80, 90, 100]

# Compute mean of vectors by stacking into matrix
X = np.array([x1, x2, x3])
mean_vector = np.mean(X, axis=0)
print("Mean of each row (axis=1):   ", np.mean(X, axis=1))
print("Mean of each column (axis=0):", mean_vector)
```

## **axis=0** down columns, **axis=1** across rows

```python
# (on the slides, this code is hidden: only its output shows)
import numpy as np, matplotlib.pyplot as plt
RED_, BLUE_ = "#d1242f", "#2a78d6"
A_ = np.array([[1, 2, 3], [4, 5, 9]])
fig, axes = plt.subplots(1, 2, figsize=(15, 4.8))
def grid(ax):
    for i in range(2):
        for j in range(3):
            ax.add_patch(plt.Rectangle((j, 1 - i), 1, 1, fill=False, lw=2, ec="#555"))
            ax.text(j + 0.4, 1.5 - i, str(A_[i, j]), ha="center", va="center", fontsize=28, weight="bold")
            ax.text(j + 0.06, 1.94 - i, f"A[{i},{j}]", ha="left", va="top", fontsize=11, color="#888")
    ax.text(-0.15, 1.5, "row 0", ha="right", va="center", fontsize=14); ax.text(-0.15, 0.5, "row 1", ha="right", va="center", fontsize=14)
    ax.set_aspect("equal"); ax.axis("off")
ax = axes[0]; grid(ax)
for j in range(3):
    ax.annotate("", xy=(j + 0.8, -0.25), xytext=(j + 0.8, 1.9), arrowprops=dict(arrowstyle="->", lw=3, color=RED_))
    ax.text(j + 0.5, -0.55, f"{A_[:, j].mean():g}", ha="center", va="center", fontsize=24, color=RED_, weight="bold")
ax.set_title("np.mean(A, axis=0): down each column", color=RED_, fontsize=17)
ax.set_xlim(-0.9, 3.4); ax.set_ylim(-0.9, 2.2)
ax = axes[1]; grid(ax)
for i in range(2):
    ax.annotate("", xy=(3.3, 1.2 - i), xytext=(0.1, 1.2 - i), arrowprops=dict(arrowstyle="->", lw=3, color=BLUE_))
    ax.text(3.65, 1.5 - i, f"{A_[i, :].mean():g}", ha="center", va="center", fontsize=24, color=BLUE_, weight="bold")
ax.set_title("np.mean(A, axis=1): across each row", color=BLUE_, fontsize=17)
ax.set_xlim(-0.9, 4.1); ax.set_ylim(-0.9, 2.2)
plt.tight_layout(); plt.show()
```

## Can we always trust the mean alone?

```python
import numpy as np

# If one number can throw off the mean this much, maybe we need something more robust.
x = [100, 105, 110, 115, 120]
x_outlier = [100, 105, 110, 115, 200]

print("Mean (normal):", np.mean(x))
print("Mean (with outlier):", np.mean(x_outlier))
```

## One outlier drags the mean along

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,3))
plt.plot(x, 'o-', label='Normal Readings')
plt.plot(x_outlier, 'o-', label='With Outlier')
plt.legend()
plt.title("Same Data, But One Outlier")
plt.xlabel("Measurement #")
plt.ylabel("Blood Pressure (mmHg)")
plt.show()
```

## Median (and Mean) of a matrix

```python
import numpy as np

data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 999]
])

print("Array:")
print(data)

print("\nMedian of all values: ", np.median(data))
print("Mean of all values:  ", np.mean(data))
print("\n*** Includes ALL values in the matrix, at once.*** ")
```

## **Median** per row, or per column

```python
# Median along rows (axis=1)
# → takes the median of each row
print("\nMedian along rows (axis=1):", np.median(data, axis=1))
```

```python
# Median along columns (axis=0)
# → takes the median of each column
print("\nMedian along columns (axis=0):", np.median(data, axis=0))
```

## Mode examples, and `statistics.mode` in Python

```python
from statistics import mode, multimode

coffee_orders = ["Latte", "Latte", "Espresso", "Mocha", "Latte", "Espresso"]
scores = [78, 85, 85, 90, 92, 92, 85, 78, 92]

print("Mode (coffee):", mode(coffee_orders))
print("Mode (scores):", mode(scores))
print("All modes (scores):", multimode(scores))
```

## Same mean, same median...

```python
import numpy as np

x1 = [98, 99, 100, 101, 102]    
x2 = [80, 90, 100, 110, 120]

mean1, mean2 = np.mean(x1), np.mean(x2)

print("Mean (x1):", mean1, " Median:", np.median(x1))
print("Mean (x2):", mean2, " Median:", np.median(x2))
```

## Sometimes, **.std** divides by $n$, sometimes by $(n-1)$?!

```python
import numpy as np
x1 = [98, 99, 100, 101, 102]
print(np.std(x1))           # divides by n
print(np.std(x1, ddof=1))   # divides by n-1, like pandas' .std()
```

## Answer

```python
import numpy as np
print("standard deviation:", np.std([2, 4, 4, 4, 5, 5, 7, 9])) #can use axis here as well!
print("variance:", np.var([2, 4, 4, 4, 5, 5, 7, 9]))
```

## The std tells them apart

```python
std1, std2 = np.std(x1), np.std(x2)

# Visualization
fig, ax = plt.subplots(figsize=(5,3))

# Plot points
ax.plot(x1, 'o-', color='steelblue', label=f"x₁ (std={std1:.1f})")
ax.plot(x2, 'o-', color='tomato', label=f"x₂ (std={std2:.1f})")

# Add mean lines
ax.axhline(mean1, color='gray', linestyle='--', alpha=0.7, label='Mean = 100')

# Add shaded regions for ±1 standard deviation
ax.fill_between(range(len(x1)),
                mean1 - std1, mean1 + std1,
                color='steelblue', alpha=0.1)
ax.fill_between(range(len(x2)),
                mean2 - std2, mean2 + std2,
                color='tomato', alpha=0.1)

ax.set_title("Same Mean, Different Standard Deviations", fontsize=10)
ax.set_xlabel("Observation index", fontsize=9)
ax.set_ylabel("Value", fontsize=9)
ax.legend(fontsize=8)
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
