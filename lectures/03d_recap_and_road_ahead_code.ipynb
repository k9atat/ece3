## Center

```python
# (on the slides, this code is hidden: only its output shows)
x         = np.array([100, 105, 110, 115, 120])
x_outlier = np.array([100, 105, 110, 115, 200])        # one bad reading
fig, axes = plt.subplots(2, 1, figsize=(15, 3.4), sharex=True)
for ax, d, label in zip(axes, [x, x_outlier], ["five readings", "the same, with one bad reading"]):
    ax.plot(d, np.zeros_like(d), "o", ms=16, color=GRAY, mec="#333")
    ax.axvline(d.mean(), color=ORANGE, lw=4, label=f"mean = {d.mean():.0f}")
    ax.axvline(np.median(d), color=BLUE, lw=4, ls="--", label=f"median = {np.median(d):.0f}")
    ax.set_yticks([]); ax.spines["left"].set_visible(False)
    ax.set_title(label, loc="left", fontsize=15); ax.legend(loc="center left", bbox_to_anchor=(1.01, 0.5), frameon=False)
axes[1].set_xlabel("reading")
plt.tight_layout(); plt.show()
```

## Spread: same mean, very different data

```python
# (on the slides, this code is hidden: only its output shows)
x1 = np.array([98, 99, 100, 101, 102])
x2 = np.array([80, 90, 100, 110, 120])
fig, axes = plt.subplots(2, 1, figsize=(15, 3.4), sharex=True)
for ax, d, label in zip(axes, [x1, x2], ["x1", "x2"]):
    m, s = d.mean(), d.std()
    ax.axvspan(m - s, m + s, color="#3fb950", alpha=0.25, label=f"mean ± std (std = {s:.1f})")
    ax.plot(d, np.zeros_like(d), "o", ms=16, color=GRAY, mec="#333")
    ax.axvline(m, color=ORANGE, lw=4, label=f"mean = {m:.0f}")
    ax.set_yticks([]); ax.spines["left"].set_visible(False)
    ax.set_title(label, loc="left", fontsize=15); ax.legend(loc="center left", bbox_to_anchor=(1.01, 0.5), frameon=False)
axes[1].set_xlabel("value")
plt.tight_layout(); plt.show()
```

## Categorical data

```python
# (on the slides, this code is hidden: only its output shows)
colors = pd.Series(["Red", "Blue", "Red", "Green", "Blue", "Red", "Blue", "Blue"])
counts = colors.value_counts()
pct = colors.value_counts(normalize=True) * 100
fig, ax = plt.subplots(figsize=(15, 3.2))
shade = {"Blue": "#2a78d6", "Red": "#d62828", "Green": "#2a9d8f"}
ax.barh(counts.index, counts.values, color=[shade[c] for c in counts.index], height=0.6)
for i, c in enumerate(counts.index):
    ax.text(counts[c] + 0.08, i, f"{counts[c]}  ({pct[c]:.1f}%)", va="center", fontsize=15)
ax.invert_yaxis(); ax.set_xlim(0, 5.2); ax.set_xlabel("count  (percent of all 8)")
plt.tight_layout(); plt.show()
```

## Correlation r

```python
# (on the slides, this code is hidden: only its output shows)
rng = np.random.default_rng(1)
xs = rng.uniform(0, 10, 60)
ys = [xs + rng.normal(0, 1.2, 60), -xs + rng.normal(0, 1.2, 60),
      rng.normal(5, 2.5, 60), (xs - 5) ** 2 + rng.normal(0, 1.5, 60)]
notes = ["strong, positive", "strong, negative", "no pattern", "a strong CURVE"]
fig, axes = plt.subplots(1, 4, figsize=(15, 3.3))
for ax, y, note in zip(axes, ys, notes):
    r = np.corrcoef(xs, y)[0, 1]
    ax.plot(xs, y, "o", ms=6, color=BLUE, alpha=0.75)
    ax.set_title(f"r = {r:+.2f}\n{note}", fontsize=15, color="#d62828" if note.endswith("CURVE") else "#333")
    ax.set_xticks([]); ax.set_yticks([])
plt.tight_layout(); plt.show()
```

## Sums of random plays

```python
# (on the slides, this code is hidden: only its output shows)
values = np.array([-2, -1, 0, 1, 2]); probs = np.array([0.3, 0.2, 0.0, 0.0, 0.5])
fig, axes = plt.subplots(1, 3, figsize=(15, 3.6))
pmf = probs
for ax, n in zip(axes, [1, 2, 30]):
    pmf = probs
    for _ in range(n - 1):
        pmf = np.convolve(pmf, probs)
    t = np.arange(-2 * n, 2 * n + 1)
    ax.vlines(t, 0, pmf, color=BLUE, lw=3 if n < 5 else 2); ax.plot(t, pmf, "o", color=BLUE, ms=4)
    ax.set_title(f"total after {n} play" + ("s" if n > 1 else "")); ax.set_yticks([])
axes[0].set_ylabel("probability")
plt.tight_layout(); plt.show()
```

## Blackjack: one deck, or an "infinite" shoe? (answer)

```python
from fractions import Fraction as F
one_deck = 2 * F(4, 52) * F(16, 51)      # without replacement
infinite = 2 * F(4, 52) * F(16, 52)      # with replacement
print(f"one deck: {float(one_deck):.4f}   infinite shoe: {float(infinite):.4f}")
```

## The binomial distribution

```python
from scipy.stats import binom
print(round(binom.pmf(2, 3, 0.5), 3))          # exactly 2 heads in 3 flips
print(round(1 - binom.pmf(0, 10, 0.048), 3))   # at least one blackjack in 10 hands
```
