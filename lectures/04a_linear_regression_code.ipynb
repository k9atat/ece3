## Why the column of ones?

```python
# (on the slides, this code is hidden: only its output shows)
Io = I + 15                      # a meter that reads 15 mA too high
Xa = np.column_stack([V, np.ones(V.size)]); Xb = V.reshape(-1, 1)
wa = np.linalg.lstsq(Xa, Io, rcond=None); wb = np.linalg.lstsq(Xb, Io, rcond=None)
fig, axes = plt.subplots(1, 2, figsize=(15, 4.6), sharey=True)
for ax, (ttl, pred, ss, col) in zip(axes, [(f"X = [V, ones]: a = {wa[0][0]:.2f}, b = {wa[0][1]:.2f}", Xa @ wa[0], wa[1][0], ORANGE),
                                         (f"X = [V] only: a = {wb[0][0]:.2f}, b forced to 0", Xb @ wb[0], wb[1][0], PURPLE)]):
    ax.plot(V, Io, "o", ms=9, color=BLUE); ax.plot(V, pred, color=col, lw=3); ax.axhline(0, color="#999", lw=1)
    ax.set_title(ttl + f"\nRSS = {ss:.1f}", loc="left"); ax.set_xlabel("voltage V (volts)"); ax.grid(alpha=0.3)
axes[0].set_ylabel("current I (mA)")
plt.tight_layout(); plt.show()
```

## In Python: `polyfit` and `lstsq`

```python
a, b = np.polyfit(V, I, 1)       # 1 = a line
print(f"a = {a:.3f} mA/V,  b = {b:.2f} mA,  R = 1000/a = {1000/a:.0f} ohm")

X = np.column_stack([V, np.ones(V.size)])
coef, ssr, rank, sv = np.linalg.lstsq(X, I, rcond=None)
print("coef:", coef)
print("ssr: ", ssr, " rank:", rank)
print(type(ssr), ssr.shape)       # an array holding one number
print(f"RMSE: {np.sqrt(ssr[0] / len(I)):.2f} mA")
```

## In Python: `minimize`

```python
from scipy.optimize import minimize

def loss(w):                  # w = [a, b]
    pred = w[0] * V + w[1]
    return np.mean((I - pred) ** 2)

w = minimize(loss, [0, 0]).x  # start: a = 0, b = 0
print(np.round(w, 2))
print(np.round(np.linalg.lstsq(X, I, rcond=None)[0], 2))
```

## Which fit has the largest R²?

```python
# (on the slides, this code is hidden: only its output shows)
rng = np.random.default_rng(8)
xx = rng.uniform(0, 10, 50)
cases = [("P", 3 * xx + rng.normal(0, 8, 50)), ("Q", 3 * xx + rng.normal(0, 1.5, 50)), ("R", rng.normal(15, 8, 50))]
fig, axes = plt.subplots(1, 3, figsize=(15, 4.2), sharex=True, sharey=True)
r2s = {}
for ax, (name, yy) in zip(axes, cases):
    sl, ic = np.polyfit(xx, yy, 1)
    r2s[name] = 1 - np.var(yy - (sl * xx + ic)) / np.var(yy)
    ax.plot(xx, yy, "o", ms=5, color=BLUE); xs = np.array([0, 10]); ax.plot(xs, sl * xs + ic, color=ORANGE, lw=3)
    ax.set_title(name, fontsize=20, weight="bold"); ax.set_xlim(0, 10); ax.set_xticks([0, 5, 10])
    ax.grid(True, color="#e3e3e3"); ax.set_axisbelow(True)
plt.tight_layout(); plt.show()
```

## Two inputs: a plane

```python
# (on the slides, this code is hidden: only its output shows)
cars = pd.read_csv("data/mpg.csv").dropna()          # 392 cars, 1970 to 1982
wt, yr, mpg = cars["weight"].to_numpy(), cars["model_year"].to_numpy() + 1900, cars["mpg"].to_numpy()
X1 = np.column_stack([wt, np.ones(len(wt))]); X2 = np.column_stack([wt, yr, np.ones(len(wt))])
def r2(X):
    res = mpg - X @ np.linalg.lstsq(X, mpg, rcond=None)[0]
    return 1 - np.var(res) / np.var(mpg)
k2 = np.linalg.lstsq(X2, mpg, rcond=None)[0]
fig = plt.figure(figsize=(15, 5.4))
ax = fig.add_subplot(1, 2, 1)
sc = ax.scatter(wt, mpg, s=12, c=yr, cmap="viridis"); ax.set_xlabel("weight (lb)"); ax.set_ylabel("mpg")
plt.colorbar(sc, ax=ax, label="model year")
ax.set_title(f"weight only: R² = {r2(X1):.2f}", loc="left")
ax3 = fig.add_subplot(1, 2, 2, projection="3d")
G1, G2 = np.meshgrid(np.linspace(1600, 5100, 10), np.linspace(1970, 1982, 10))
ax3.plot_surface(G1, G2, k2[0] * G1 + k2[1] * G2 + k2[2], alpha=0.3, color="red", edgecolor="none")
ax3.scatter(wt, yr, mpg, s=6, c=yr, cmap="viridis", depthshade=False)
ax3.set_xlabel("weight (lb)", labelpad=12); ax3.set_ylabel("model year", labelpad=12); ax3.set_zlabel("mpg"); ax3.view_init(18, -60)
ax3.set_title(f"weight + model year, a plane: R² = {r2(X2):.2f}")
plt.tight_layout(); plt.show()
```

## Physics for Lab 4: a rock, and a rocket

```python
# (on the slides, this code is hidden: only its output shows)
fig, axes = plt.subplots(1, 2, figsize=(14, 5.0))
def body(ax, x, label, thrust):
    ax.add_patch(plt.Rectangle((x - 0.5, 2.0), 1.0, 2.2, fc="#3a3550" if thrust else "#8a8070", ec="k"))
    ax.annotate("", xy=(x, 0.4), xytext=(x, 2.0), arrowprops=dict(arrowstyle="-|>", lw=4, color=RED, mutation_scale=28))
    ax.text(x + 0.15, 0.9, "weight  m g", color=RED, fontsize=17, va="center")
    if thrust:
        ax.annotate("", xy=(x, 6.0), xytext=(x, 4.2), arrowprops=dict(arrowstyle="-|>", lw=4, color=GREEN, mutation_scale=28))
        ax.text(x + 0.15, 5.2, "thrust F", color=GREEN, fontsize=17, va="center")
        ax.fill([x - 0.3, x, x + 0.3], [2.0, 1.2, 2.0], color="orange", zorder=1)          # the engine's flame
    ax.text(x, 6.9, label, ha="center", fontsize=19, weight="bold")
for ax, (label, thrust, eq) in zip(axes, [("a rock: gravity only", False, "a = −g = −9.81 m/s²"),
                                          ("a rocket: gravity + engine", True, "a = F/m − g")]):
    ax.set_xlim(-3, 3); ax.set_ylim(0, 7.4); ax.axis("off")
    body(ax, 0.6, label, thrust)
    ax.text(0.6, -0.2, eq, ha="center", fontsize=21, color="#222")
plt.tight_layout(); plt.show()
```
